````python
# ============================================================
# CELL 8 — DATA LEAKAGE CHECK
# ============================================================

def get_chatml_user_input(example):
    """
    Extract the actual customer ticket from a ChatML example.
    """
    for message in example["messages"]:
        if message["role"] == "user":
            content = message["content"]

            # Our training format is:
            # Customer Ticket:
            # <actual ticket>
            if "Customer Ticket:" in content:
                return content.split(
                    "Customer Ticket:",
                    1
                )[1].strip()

            return content.strip()

    return None


train_all_inputs = set(
    get_chatml_user_input(x)
    for x in chatml
)

annotation_inputs = set(
    str(x["input"]).strip()
    for x in annotations
)

benchmark_inputs = set(
    benchmark_df["input"]
    .astype(str)
    .str.strip()
)

original_inputs = set(
    original_df["input"]
    .astype(str)
    .str.strip()
)

print("=" * 70)
print("DATA LEAKAGE AUDIT")
print("=" * 70)

print(
    "ChatML ↔ Annotations:",
    len(train_all_inputs & annotation_inputs)
)

print(
    "ChatML ↔ Benchmark:",
    len(train_all_inputs & benchmark_inputs)
)

print(
    "Annotations ↔ Benchmark:",
    len(annotation_inputs & benchmark_inputs)
)

print(
    "Benchmark tickets:",
    len(benchmark_inputs)
)

print(
    "Training tickets:",
    len(train_all_inputs)
)

# ------------------------------------------------------------
# HARD FAIL CONDITIONS
# ------------------------------------------------------------

if train_all_inputs & benchmark_inputs:
    raise RuntimeError(
        "❌ TRAINING ↔ BENCHMARK LEAKAGE DETECTED"
    )

if annotation_inputs & benchmark_inputs:
    raise RuntimeError(
        "❌ ANNOTATION ↔ BENCHMARK LEAKAGE DETECTED"
    )

if len(train_all_inputs) != len(chatml):
    raise RuntimeError(
        "❌ DUPLICATE INPUTS EXIST IN CHATML DATA"
    )

print()
print("✓ NO TRAINING/BENCHMARK LEAKAGE")
print("✓ NO DUPLICATE TRAINING INPUTS")
````

````python
# ============================================================
# CELL 15 — TOKEN LENGTH CHECK
# ============================================================

import numpy as np

lengths = []

for example in train_dataset:

    text_parts = []

    for message in example["messages"]:
        text_parts.append(
            f"{message['role']}: {message['content']}"
        )

    text = "\n".join(text_parts)

    tokens = tokenizer(
        text,
        add_special_tokens=True,
        truncation=False,
    )

    lengths.append(
        len(tokens["input_ids"])
    )

lengths = np.array(lengths)

print("=" * 70)
print("TOKEN LENGTH ANALYSIS")
print("=" * 70)

print("Samples:", len(lengths))
print("Minimum:", lengths.min())
print("Mean   :", round(lengths.mean(), 1))
print("Median :", np.median(lengths))
print("95th % :", np.percentile(lengths, 95))
print("99th % :", np.percentile(lengths, 99))
print("Maximum:", lengths.max())

print()
print(
    "Exceeding 2048:",
    int((lengths > MAX_SEQ_LENGTH).sum())
)
````

````python
# ============================================================
# CELL 18 — PREPARE TEXT DATASET
# ============================================================

from datasets import Dataset

print("=" * 70)
print("CONVERTING CHATML → TEXT")
print("=" * 70)


def convert_messages_to_text(example):
    """
    Convert the existing messages structure into one plain
    string using the model's native chat template.
    """

    text = tokenizer.apply_chat_template(
        example["messages"],
        tokenize=False,
        add_generation_prompt=False,
    )

    return {
        "text": text
    }


train_text_dataset = train_dataset.map(
    convert_messages_to_text,
    remove_columns=train_dataset.column_names,
    desc="Formatting training dataset",
)

validation_text_dataset = validation_dataset.map(
    convert_messages_to_text,
    remove_columns=validation_dataset.column_names,
    desc="Formatting validation dataset",
)


print()
print("Training examples   :", len(train_text_dataset))
print("Validation examples :", len(validation_text_dataset))

print()
print("Training columns:")
print(train_text_dataset.column_names)

print()
print("Text type:")
print(type(train_text_dataset[0]["text"]))

print()
print("First formatted example:")
print(train_text_dataset[0]["text"])
````

````python
# ============================================================
# CELL 26 — 5-TICKET SMOKE TEST
# ============================================================

import json
import re
import torch

model.eval()

test_tickets = benchmark_df["input"].head(5).tolist()

def extract_json(text):
    """
    Extract the first JSON object from model output.
    """
    match = re.search(
        r"\{.*\}",
        text,
        re.DOTALL
    )

    if not match:
        return None

    try:
        return json.loads(match.group(0))
    except json.JSONDecodeError:
        return None


def predict_ticket(ticket):

    messages = [
        {
            "role": "system",
            "content": (
                "You are TicketMindAI, a customer support "
                "ticket classification system.\n\n"
                "Analyze the customer's ticket and return "
                "structured metadata.\n\n"
                "You must determine:\n"
                "- category\n"
                "- subcategory\n"
                "- priority\n"
                "- sentiment\n"
                "- summary\n\n"
                "Return ONLY valid JSON."
            ),
        },
        {
            "role": "user",
            "content": f"Customer Ticket:\n{ticket}",
        },
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=MAX_SEQ_LENGTH,
        padding=False,
    )

    inputs = {
        k: v.to(model.device)
        for k, v in inputs.items()
    }

    with torch.inference_mode():

        outputs = model.generate(
            **inputs,
            max_new_tokens=128,
            do_sample=False,
            temperature=None,
            top_p=None,
            pad_token_id=tokenizer.eos_token_id,
        )

    generated_tokens = outputs[
        0,
        inputs["input_ids"].shape[1]:
    ]

    generated_text = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True,
    )

    return generated_text


print("=" * 70)
print("5-TICKET SMOKE TEST")
print("=" * 70)

valid_json = 0

for i, ticket in enumerate(test_tickets, 1):

    raw = predict_ticket(ticket)

    parsed = extract_json(raw)

    print()
    print(f"--- TEST {i} ---")
    print("INPUT:", ticket)
    print("RAW:", raw)

    if parsed is not None:

        valid_json += 1

        print("JSON: ✓")

    else:

        print("JSON: ✗")


print()
print("=" * 70)
print(
    f"VALID JSON: {valid_json}/5"
)
print("=" * 70)
````

````python
# ============================================================
# TICKETMINDAI — FINAL QLoRA 540-TICKET BENCHMARK
# ============================================================
#
# PURPOSE:
#   Compare the fine-tuned QLoRA TicketMindAI model against
#   the established prompt baseline on the EXACT same 540
#   benchmark samples.
#
# IMPORTANT:
#   - Deterministic generation
#   - Exact benchmark order
#   - Exact system prompt
#   - No label repair
#   - Invalid priorities remain INVALID
#   - Per-ticket persistent checkpointing
#   - Checkpoint stored on Google Drive when available
#   - Automatically resumes if interrupted
#   - Final metrics + predictions + confusion matrices saved
#
# ============================================================

import os
import re
import json
import ast
import shutil
import hashlib

import pandas as pd
import numpy as np
import torch

from tqdm.auto import tqdm
from unsloth import FastLanguageModel

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
)

# ============================================================
# 1. GLOBAL CONFIGURATION
# ============================================================

SEED = 42

TOTAL_BENCHMARK_ROWS = 540

MAX_SEQ_LENGTH = 512
MAX_NEW_TOKENS = 128

N_REPRESENTATIVE_ERRORS = 15

VALID_PRIORITIES = {
    "HIGH",
    "MEDIUM",
    "LOW",
}

# ============================================================
# 2. GOOGLE DRIVE / COLAB PATH RESOLUTION
# ============================================================
#
# We prefer Google Drive for all persistent outputs.
#
# /content can disappear if the Colab runtime dies.
# Google Drive survives runtime resets.
# ============================================================

DRIVE_PROJECT_ROOT = (
    "/content/drive/MyDrive/TicketMindAI"
)

LOCAL_PROJECT_ROOT = (
    "/content"
)

if os.path.isdir(DRIVE_PROJECT_ROOT):

    PROJECT_ROOT = DRIVE_PROJECT_ROOT

    print(
        "✓ Google Drive project detected."
    )

else:

    PROJECT_ROOT = LOCAL_PROJECT_ROOT

    print(
        "⚠ Google Drive project not detected."
    )

    print(
        "Using /content for this runtime."
    )

# ============================================================
# 3. INPUT / OUTPUT PATHS
# ============================================================

BENCHMARK_CANDIDATES = [

    os.path.join(
        PROJECT_ROOT,
        "data",
        "processed",
        "benchmark.csv",
    ),

    "/content/data/processed/benchmark.csv",

    "/content/drive/MyDrive/TicketMindAI/data/processed/benchmark.csv",
]

ADAPTER_CANDIDATES = [

    os.path.join(
        PROJECT_ROOT,
        "outputs",
        "ticketmind-qlora",
        "final_adapter",
    ),

    "/content/outputs/ticketmind-qlora/final_adapter",

    "/content/drive/MyDrive/TicketMindAI/outputs/ticketmind-qlora/final_adapter",
]

# ============================================================
# 4. FIND BENCHMARK
# ============================================================

BENCHMARK_PATH = None

for candidate in BENCHMARK_CANDIDATES:

    if os.path.isfile(candidate):

        BENCHMARK_PATH = candidate

        break

if BENCHMARK_PATH is None:

    raise FileNotFoundError(
        "Could not find benchmark.csv.\n\n"
        "Checked:\n"
        + "\n".join(BENCHMARK_CANDIDATES)
    )

# ============================================================
# 5. FIND QLORA ADAPTER
# ============================================================

MODEL_PATH = None

for candidate in ADAPTER_CANDIDATES:

    if os.path.isdir(candidate):

        config_file = os.path.join(
            candidate,
            "adapter_config.json",
        )

        weights_file = os.path.join(
            candidate,
            "adapter_model.safetensors",
        )

        if (
            os.path.isfile(config_file)
            and
            os.path.isfile(weights_file)
        ):

            MODEL_PATH = candidate

            break

if MODEL_PATH is None:

    raise FileNotFoundError(
        "Could not find a valid QLoRA adapter.\n\n"
        "Required files:\n"
        "  adapter_config.json\n"
        "  adapter_model.safetensors\n\n"
        "Checked:\n"
        + "\n".join(ADAPTER_CANDIDATES)
    )

# ============================================================
# 6. OUTPUT DIRECTORY
# ============================================================

OUT_DIR = os.path.join(
    PROJECT_ROOT,
    "outputs",
    "ticketmind-qlora",
)

os.makedirs(
    OUT_DIR,
    exist_ok=True,
)

CONFUSION_DIR = os.path.join(
    OUT_DIR,
    "confusion_matrices",
)

os.makedirs(
    CONFUSION_DIR,
    exist_ok=True,
)

# ============================================================
# 7. IMPORTANT:
#    USE A NEW CHECKPOINT FOR THIS FINAL RUN
# ============================================================
#
# This prevents an older experimental checkpoint from being
# accidentally mixed with this final benchmark implementation.
#
# If this exact run gets interrupted, THIS SAME FILE allows
# resumption.
# ============================================================

CHECKPOINT_PATH = os.path.join(
    OUT_DIR,
    "FINAL_540_checkpoint.jsonl",
)

RAW_PREDICTIONS_PATH = os.path.join(
    OUT_DIR,
    "FINAL_540_predictions.json",
)

METRICS_PATH = os.path.join(
    OUT_DIR,
    "FINAL_540_metrics.json",
)

ERRORS_PATH = os.path.join(
    OUT_DIR,
    "FINAL_540_representative_errors.csv",
)

# ============================================================
# 8. EXACT SYSTEM PROMPT
# ============================================================

SYSTEM_PROMPT = """You are TicketMindAI, a customer support ticket classification system.

Analyze the customer's ticket and return structured metadata.

You must determine:
- category
- subcategory
- priority
- sentiment
- summary

Return ONLY valid JSON.

Required format:
{
  "category": "...",
  "subcategory": "...",
  "priority": "HIGH|MEDIUM|LOW",
  "sentiment": "Positive|Neutral|Negative",
  "summary": "..."
}"""

# ============================================================
# 9. EXISTING PROMPT BASELINE
# ============================================================

BASELINE_RESULTS = {

    "category": {
        "accuracy": 0.9741,
        "macro_precision": 0.9762,
        "macro_recall": 0.9708,
        "macro_f1": 0.9716,
    },

    "subcategory": {
        "accuracy": 0.9481,
        "macro_precision": 0.9537,
        "macro_recall": 0.9481,
        "macro_f1": 0.9480,
    },

    "priority": {
        "accuracy": 0.3870,
        "macro_precision": 0.2176,
        "macro_recall": 0.3653,
        "macro_f1": 0.2711,
    },
}

# ============================================================
# 10. REPRODUCIBILITY
# ============================================================

torch.manual_seed(SEED)

np.random.seed(SEED)

if torch.cuda.is_available():

    torch.cuda.manual_seed_all(SEED)

# ============================================================
# 11. ENVIRONMENT CHECK
# ============================================================

print()
print("=" * 70)
print("TICKETMINDAI — FINAL QLoRA BENCHMARK")
print("=" * 70)

print(
    "Benchmark:",
    BENCHMARK_PATH,
)

print(
    "Adapter:",
    MODEL_PATH,
)

print(
    "Checkpoint:",
    CHECKPOINT_PATH,
)

print(
    "Python:",
    os.sys.version.split()[0],
)

print(
    "PyTorch:",
    torch.__version__,
)

print(
    "CUDA:",
    torch.version.cuda,
)

print(
    "CUDA available:",
    torch.cuda.is_available(),
)

if not torch.cuda.is_available():

    raise RuntimeError(
        "CUDA GPU is unavailable.\n"
        "Enable Colab GPU before running this benchmark."
    )

print(
    "GPU:",
    torch.cuda.get_device_name(0),
)

print(
    "VRAM:",
    round(
        torch.cuda.get_device_properties(0)
        .total_memory
        / 1024**3,
        2,
    ),
    "GB",
)

# ============================================================
# 12. VALIDATE ADAPTER
# ============================================================

ADAPTER_CONFIG = os.path.join(
    MODEL_PATH,
    "adapter_config.json",
)

ADAPTER_WEIGHTS = os.path.join(
    MODEL_PATH,
    "adapter_model.safetensors",
)

assert os.path.isfile(
    ADAPTER_CONFIG
)

assert os.path.isfile(
    ADAPTER_WEIGHTS
)

print()
print(
    "✓ adapter_config.json"
)

print(
    "✓ adapter_model.safetensors"
)

# ============================================================
# 13. LOAD BENCHMARK
# ============================================================

df = pd.read_csv(
    BENCHMARK_PATH
).reset_index(drop=True)

assert len(df) == (
    TOTAL_BENCHMARK_ROWS
), (
    f"Expected {TOTAL_BENCHMARK_ROWS} rows, "
    f"found {len(df)}"
)

required_columns = {

    "input",
    "category",
    "subcategory",
    "priority",
}

missing_columns = (
    required_columns
    -
    set(df.columns)
)

if missing_columns:

    raise RuntimeError(
        "Benchmark is missing columns: "
        f"{missing_columns}"
    )

# ============================================================
# 14. PERMANENT BENCHMARK INDEX
# ============================================================

df["benchmark_index"] = np.arange(
    len(df)
)

# ============================================================
# 15. NORMALIZE ONLY GROUND-TRUTH PRIORITY
# ============================================================

df["category"] = (
    df["category"]
    .astype(str)
    .str.strip()
)

df["subcategory"] = (
    df["subcategory"]
    .astype(str)
    .str.strip()
)

df["priority"] = (
    df["priority"]
    .astype(str)
    .str.strip()
    .str.upper()
)

CATEGORY_CLASSES = sorted(
    df["category"]
    .unique()
    .tolist()
)

SUBCATEGORY_CLASSES = sorted(
    df["subcategory"]
    .unique()
    .tolist()
)

print()
print("=" * 70)
print("BENCHMARK VALIDATION")
print("=" * 70)

print(
    "Rows:",
    len(df),
)

print(
    "Categories:",
    len(CATEGORY_CLASSES),
)

print(
    "Subcategories:",
    len(SUBCATEGORY_CLASSES),
)

# ============================================================
# 16. LOAD MODEL
# ============================================================

print()
print("=" * 70)
print("LOADING QLoRA MODEL")
print("=" * 70)

model, tokenizer = (
    FastLanguageModel.from_pretrained(
        model_name=MODEL_PATH,
        max_seq_length=MAX_SEQ_LENGTH,
        load_in_4bit=True,
    )
)

FastLanguageModel.for_inference(
    model
)

# ============================================================
# 17. GENERATION CONFIG
# ============================================================
#
# Critical:
# max_length from Qwen's generation_config can conflict with
# max_new_tokens.
#
# We explicitly remove it.
# ============================================================

model.generation_config.max_length = None

model.generation_config.max_new_tokens = (
    MAX_NEW_TOKENS
)

model.generation_config.do_sample = (
    False
)

model.generation_config.temperature = (
    None
)

model.generation_config.top_p = (
    None
)

model.generation_config.top_k = (
    None
)

# ============================================================
# 18. TOKENIZER
# ============================================================

if tokenizer.pad_token_id is None:

    tokenizer.pad_token = (
        tokenizer.eos_token
    )

print()
print(
    "✓ Model loaded"
)

print(
    "✓ Unsloth inference mode enabled"
)

print(
    "✓ Deterministic generation"
)

print(
    "✓ max_length disabled"
)

print(
    "✓ max_new_tokens:",
    MAX_NEW_TOKENS,
)

print(
    "✓ pad_token_id:",
    tokenizer.pad_token_id,
)

# ============================================================
# 19. DEVICE
# ============================================================
#
# DO NOT call model.to("cuda") here.
#
# For 4-bit Unsloth models, device placement is handled by
# the model loader.
# ============================================================

DEVICE = "cuda"

# ============================================================
# 20. JSON EXTRACTION
# ============================================================

def extract_json(text):

    if text is None:

        return None

    text = str(text).strip()

    if not text:

        return None

    # Remove markdown JSON fences.

    text = re.sub(
        r"^```json\s*",
        "",
        text,
        flags=re.IGNORECASE,
    )

    text = re.sub(
        r"^```\s*",
        "",
        text,
    )

    text = re.sub(
        r"\s*```$",
        "",
        text,
    )

    text = text.strip()

    # Exact JSON first.

    try:

        return json.loads(text)

    except json.JSONDecodeError:

        pass

    # Embedded JSON object.

    match = re.search(
        r"\{.*\}",
        text,
        flags=re.DOTALL,
    )

    if match:

        candidate = (
            match.group(0)
        )

        try:

            return json.loads(
                candidate
            )

        except json.JSONDecodeError:

            pass

        # Python-style dict fallback.

        try:

            return ast.literal_eval(
                candidate
            )

        except (
            ValueError,
            SyntaxError,
        ):

            pass

    return None

# ============================================================
# 21. NORMALIZE MODEL OUTPUT
# ============================================================

def normalize_prediction(
    raw_prediction
):

    if not isinstance(
        raw_prediction,
        dict,
    ):

        return None

    category = str(
        raw_prediction.get(
            "category",
            "",
        )
    ).strip()

    subcategory = str(
        raw_prediction.get(
            "subcategory",
            "",
        )
    ).strip()

    sentiment = str(
        raw_prediction.get(
            "sentiment",
            "",
        )
    ).strip()

    summary = str(
        raw_prediction.get(
            "summary",
            "",
        )
    ).strip()

    priority_raw = (
        raw_prediction.get(
            "priority",
            "",
        )
    )

    priority_raw_str = str(
        priority_raw
    ).strip()

    priority_upper = (
        priority_raw_str.upper()
    )

    # ONLY canonical textual priorities
    # are considered valid.
    #
    # "High" -> HIGH
    # "LOW"  -> LOW
    #
    # 10 -> INVALID
    # 4  -> INVALID

    if priority_upper in (
        VALID_PRIORITIES
    ):

        priority_norm = (
            priority_upper
        )

        priority_valid = True

    else:

        priority_norm = (
            "INVALID"
        )

        priority_valid = False

    return {

        "category":
            category,

        "subcategory":
            subcategory,

        "priority_raw":
            priority_raw_str,

        "priority_norm":
            priority_norm,

        "priority_valid":
            priority_valid,

        "sentiment":
            sentiment,

        "summary":
            summary,
    }

# ============================================================
# 22. CHECKPOINT SCHEMA
# ============================================================

REQUIRED_RECORD_KEYS = {

    "benchmark_index",

    "input",

    "true_category",

    "true_subcategory",

    "true_priority",

    "raw_model_output",

    "valid_json",

    "malformed",

    "pred_category",

    "pred_subcategory",

    "pred_priority_raw",

    "pred_priority_norm",

    "priority_valid",

    "pred_sentiment",

    "pred_summary",
}

# ============================================================
# 23. LOAD CHECKPOINT
# ============================================================

def load_checkpoint(
    path
):

    completed = {}

    if not os.path.isfile(
        path
    ):

        return completed

    with open(
        path,
        "r",
        encoding="utf-8",
    ) as f:

        for line in f:

            line = line.strip()

            if not line:

                continue

            try:

                record = json.loads(
                    line
                )

            except json.JSONDecodeError:

                # Ignore incomplete final line.

                continue

            if not isinstance(
                record,
                dict,
            ):

                continue

            if not REQUIRED_RECORD_KEYS.issubset(
                record.keys()
            ):

                continue

            try:

                idx = int(
                    record[
                        "benchmark_index"
                    ]
                )

            except (
                TypeError,
                ValueError,
            ):

                continue

            if not (
                0 <= idx
                < TOTAL_BENCHMARK_ROWS
            ):

                continue

            completed[idx] = (
                record
            )

    return completed

# ============================================================
# 24. CHECKPOINT STATUS
# ============================================================

completed_records = (
    load_checkpoint(
        CHECKPOINT_PATH
    )
)

completed_indices = set(
    completed_records.keys()
)

remaining_df = df[
    ~df["benchmark_index"].isin(
        completed_indices
    )
].copy()

print()
print("=" * 70)
print("CHECKPOINT STATUS")
print("=" * 70)

print(
    "Total:",
    TOTAL_BENCHMARK_ROWS,
)

print(
    "Already completed:",
    len(completed_indices),
)

print(
    "Remaining:",
    len(remaining_df),
)

# ============================================================
# 25. CHECKPOINT INTEGRITY
# ============================================================

invalid_indices = (
    completed_indices
    -
    set(
        range(
            TOTAL_BENCHMARK_ROWS
        )
    )
)

if invalid_indices:

    raise RuntimeError(
        "Invalid checkpoint indices: "
        f"{sorted(invalid_indices)}"
    )

# ============================================================
# 26. OPEN CHECKPOINT
# ============================================================

checkpoint_file = open(
    CHECKPOINT_PATH,
    "a",
    encoding="utf-8",
)

# ============================================================
# 27. INFERENCE LOOP
# ============================================================

try:

    progress = tqdm(
        remaining_df.itertuples(
            index=False
        ),
        total=len(
            remaining_df
        ),
        desc=(
            "QLoRA 540 benchmark"
        ),
    )

    for row in progress:

        benchmark_index = int(
            row.benchmark_index
        )

        ticket = str(
            row.input
        )

        # ----------------------------------------------------
        # BUILD CHAT MESSAGES
        # ----------------------------------------------------

        messages = [

            {
                "role": "system",
                "content":
                    SYSTEM_PROMPT,
            },

            {
                "role": "user",
                "content":
                    (
                        "Customer Ticket:\n"
                        f"{ticket}"
                    ),
            },
        ]

        # ----------------------------------------------------
        # APPLY CHAT TEMPLATE AS TEXT
        # ----------------------------------------------------
        #
        # We do NOT tokenize=True here.
        # This avoids double-tokenization / mask mismatch.
        # ----------------------------------------------------

        prompt_text = (
            tokenizer.apply_chat_template(
                messages,
                tokenize=False,
                add_generation_prompt=True,
            )
        )

        if not isinstance(
            prompt_text,
            str,
        ):

            raise RuntimeError(
                "Chat template did not return "
                "a string."
            )

        # ----------------------------------------------------
        # TOKENIZE EXACTLY ONCE
        # ----------------------------------------------------

        encoded = tokenizer(
            prompt_text,

            return_tensors="pt",

            add_special_tokens=False,

            truncation=True,

            max_length=MAX_SEQ_LENGTH,
        )

        input_ids = (
            encoded[
                "input_ids"
            ]
            .to(DEVICE)
        )

        attention_mask = (
            encoded[
                "attention_mask"
            ]
            .to(DEVICE)
        )

        # ----------------------------------------------------
        # GENERATE
        # ----------------------------------------------------

        with torch.inference_mode():

            output_ids = (
                model.generate(

                    input_ids=input_ids,

                    attention_mask=(
                        attention_mask
                    ),

                    max_new_tokens=(
                        MAX_NEW_TOKENS
                    ),

                    do_sample=False,

                    pad_token_id=(
                        tokenizer.pad_token_id
                    ),
                )
            )

        # ----------------------------------------------------
        # REMOVE PROMPT TOKENS
        # ----------------------------------------------------

        generated_tokens = (
            output_ids[0][
                input_ids.shape[-1]:
            ]
        )

        # ----------------------------------------------------
        # DECODE ONLY GENERATED TEXT
        # ----------------------------------------------------

        raw_text = (
            tokenizer.decode(
                generated_tokens,
                skip_special_tokens=True,
            )
            .strip()
        )

        # ----------------------------------------------------
        # PARSE
        # ----------------------------------------------------

        parsed = extract_json(
            raw_text
        )

        valid_json = (
            parsed is not None
        )

        normalized = (
            normalize_prediction(
                parsed
            )
            if valid_json
            else None
        )

        malformed = (
            normalized is None
        )

        # ----------------------------------------------------
        # REPRESENT MALFORMED OUTPUT
        # WITHOUT REPAIRING IT
        # ----------------------------------------------------

        if normalized is None:

            normalized = {

                "category":
                    "",

                "subcategory":
                    "",

                "priority_raw":
                    "",

                "priority_norm":
                    "INVALID",

                "priority_valid":
                    False,

                "sentiment":
                    "",

                "summary":
                    "",
            }

        # ----------------------------------------------------
        # GROUND TRUTH
        # ----------------------------------------------------

        true_category = str(
            row.category
        ).strip()

        true_subcategory = str(
            row.subcategory
        ).strip()

        true_priority = (
            str(row.priority)
            .strip()
            .upper()
        )

        # ----------------------------------------------------
        # COMPLETE RECORD
        # ----------------------------------------------------

        record = {

            "benchmark_index":
                benchmark_index,

            "input":
                ticket,

            "true_category":
                true_category,

            "true_subcategory":
                true_subcategory,

            "true_priority":
                true_priority,

            "raw_model_output":
                raw_text,

            "valid_json":
                bool(valid_json),

            "malformed":
                bool(malformed),

            "pred_category":
                normalized[
                    "category"
                ],

            "pred_subcategory":
                normalized[
                    "subcategory"
                ],

            "pred_priority_raw":
                normalized[
                    "priority_raw"
                ],

            "pred_priority_norm":
                normalized[
                    "priority_norm"
                ],

            "priority_valid":
                bool(
                    normalized[
                        "priority_valid"
                    ]
                ),

            "pred_sentiment":
                normalized[
                    "sentiment"
                ],

            "pred_summary":
                normalized[
                    "summary"
                ],
        }

        # ----------------------------------------------------
        # WRITE IMMEDIATELY
        # ----------------------------------------------------
        #
        # Even a malformed prediction is considered a completed
        # benchmark item.
        #
        # This means we NEVER silently retry only because the
        # answer was bad.
        # ----------------------------------------------------

        checkpoint_file.write(
            json.dumps(
                record,
                ensure_ascii=False,
            )
            + "\n"
        )

        checkpoint_file.flush()

        os.fsync(
            checkpoint_file.fileno()
        )

finally:

    checkpoint_file.close()

# ============================================================
# 28. RELOAD FINAL CHECKPOINT
# ============================================================

all_records = (
    load_checkpoint(
        CHECKPOINT_PATH
    )
)

all_indices = set(
    all_records.keys()
)

expected_indices = set(
    range(
        TOTAL_BENCHMARK_ROWS
    )
)

missing_indices = sorted(
    expected_indices
    -
    all_indices
)

extra_indices = sorted(
    all_indices
    -
    expected_indices
)

# ============================================================
# 29. FINAL COMPLETENESS CHECK
# ============================================================

if extra_indices:

    raise RuntimeError(
        "Unexpected checkpoint indices:\n"
        f"{extra_indices}"
    )

if missing_indices:

    raise RuntimeError(
        "Benchmark did not finish.\n"
        f"Completed: {len(all_indices)}/"
        f"{TOTAL_BENCHMARK_ROWS}\n"
        f"Missing: {missing_indices}"
    )

assert len(
    all_records
) == TOTAL_BENCHMARK_ROWS

# ============================================================
# 30. SORT BY ORIGINAL BENCHMARK ORDER
# ============================================================

records = [

    all_records[i]

    for i in sorted(
        all_indices
    )

]

# ============================================================
# 31. SAVE FINAL PREDICTIONS
# ============================================================

with open(
    RAW_PREDICTIONS_PATH,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        records,
        f,
        indent=2,
        ensure_ascii=False,
    )

# ============================================================
# 32. DATAFRAME
# ============================================================

results_df = pd.DataFrame(
    records
)

assert len(
    results_df
) == TOTAL_BENCHMARK_ROWS

# ============================================================
# 33. METRICS FUNCTION
# ============================================================

def calculate_metrics(
    y_true,
    y_pred,
    labels,
):

    accuracy = accuracy_score(
        y_true,
        y_pred,
    )

    precision, recall, f1, _ = (
        precision_recall_fscore_support(

            y_true,

            y_pred,

            labels=labels,

            average="macro",

            zero_division=0,
        )
    )

    return {

        "accuracy":
            float(accuracy),

        "macro_precision":
            float(precision),

        "macro_recall":
            float(recall),

        "macro_f1":
            float(f1),
    }

# ============================================================
# 34. CATEGORY
# ============================================================

category_metrics = (
    calculate_metrics(

        results_df[
            "true_category"
        ],

        results_df[
            "pred_category"
        ],

        CATEGORY_CLASSES,
    )
)

# ============================================================
# 35. SUBCATEGORY
# ============================================================

subcategory_metrics = (
    calculate_metrics(

        results_df[
            "true_subcategory"
        ],

        results_df[
            "pred_subcategory"
        ],

        SUBCATEGORY_CLASSES,
    )
)

# ============================================================
# 36. PRIORITY
# ============================================================

priority_labels = [
    "HIGH",
    "MEDIUM",
    "LOW",
]

priority_metrics = (
    calculate_metrics(

        results_df[
            "true_priority"
        ],

        results_df[
            "pred_priority_norm"
        ],

        priority_labels,
    )
)

# ============================================================
# 37. JSON QUALITY
# ============================================================

valid_json_rate = float(
    results_df[
        "valid_json"
    ].mean()
)

malformed_count = int(
    results_df[
        "malformed"
    ].sum()
)

invalid_priority_count = int(
    (
        ~results_df[
            "priority_valid"
        ]
    ).sum()
)

# ============================================================
# 38. EXACT STRUCTURED MATCH
# ============================================================
#
# Benchmark ground truth contains:
#   category
#   subcategory
#   priority
#
# Therefore exact structured match is based on those 3 only.
#
# Sentiment and summary are NOT included because benchmark.csv
# does not contain their ground truth.
# ============================================================

exact_match_mask = (

    results_df[
        "true_category"
    ]
    ==
    results_df[
        "pred_category"
    ]

) & (

    results_df[
        "true_subcategory"
    ]
    ==
    results_df[
        "pred_subcategory"
    ]

) & (

    results_df[
        "true_priority"
    ]
    ==
    results_df[
        "pred_priority_norm"
    ]
)

exact_structured_match_rate = float(
    exact_match_mask.mean()
)

# ============================================================
# 39. DISTRIBUTIONS
# ============================================================

distributions = {

    "true_category":
        results_df[
            "true_category"
        ]
        .value_counts()
        .to_dict(),

    "pred_category":
        results_df[
            "pred_category"
        ]
        .value_counts()
        .to_dict(),

    "true_subcategory":
        results_df[
            "true_subcategory"
        ]
        .value_counts()
        .to_dict(),

    "pred_subcategory":
        results_df[
            "pred_subcategory"
        ]
        .value_counts()
        .to_dict(),

    "true_priority":
        results_df[
            "true_priority"
        ]
        .value_counts()
        .to_dict(),

    "pred_priority_raw":
        results_df[
            "pred_priority_raw"
        ]
        .value_counts()
        .to_dict(),

    "pred_priority_norm":
        results_df[
            "pred_priority_norm"
        ]
        .value_counts()
        .to_dict(),
}

# ============================================================
# 40. CATEGORY CONFUSION MATRIX
# ============================================================

category_cm = confusion_matrix(

    results_df[
        "true_category"
    ],

    results_df[
        "pred_category"
    ],

    labels=CATEGORY_CLASSES,
)

pd.DataFrame(

    category_cm,

    index=CATEGORY_CLASSES,

    columns=CATEGORY_CLASSES,

).to_csv(

    os.path.join(
        CONFUSION_DIR,
        "category_confusion.csv",
    )
)

# ============================================================
# 41. SUBCATEGORY CONFUSION MATRIX
# ============================================================

subcategory_cm = confusion_matrix(

    results_df[
        "true_subcategory"
    ],

    results_df[
        "pred_subcategory"
    ],

    labels=SUBCATEGORY_CLASSES,
)

pd.DataFrame(

    subcategory_cm,

    index=SUBCATEGORY_CLASSES,

    columns=SUBCATEGORY_CLASSES,

).to_csv(

    os.path.join(
        CONFUSION_DIR,
        "subcategory_confusion.csv",
    )
)

# ============================================================
# 42. PRIORITY CONFUSION MATRIX
# ============================================================

priority_cm_labels = [

    "HIGH",
    "MEDIUM",
    "LOW",
    "INVALID",
]

priority_cm = confusion_matrix(

    results_df[
        "true_priority"
    ],

    results_df[
        "pred_priority_norm"
    ],

    labels=priority_cm_labels,
)

pd.DataFrame(

    priority_cm,

    index=priority_cm_labels,

    columns=priority_cm_labels,

).to_csv(

    os.path.join(
        CONFUSION_DIR,
        "priority_confusion.csv",
    )
)

# ============================================================
# 43. REPRESENTATIVE ERRORS
# ============================================================

error_df = results_df[
    ~exact_match_mask
].copy()

if len(error_df) > (
    N_REPRESENTATIVE_ERRORS
):

    representative_errors = (
        error_df.sample(
            n=N_REPRESENTATIVE_ERRORS,
            random_state=SEED,
        )
    )

else:

    representative_errors = (
        error_df
    )

representative_errors[

    [
        "benchmark_index",

        "input",

        "true_category",
        "pred_category",

        "true_subcategory",
        "pred_subcategory",

        "true_priority",

        "pred_priority_raw",

        "pred_priority_norm",

        "valid_json",

        "malformed",
    ]

].to_csv(

    ERRORS_PATH,

    index=False,
)

# ============================================================
# 44. BASELINE DELTAS
# ============================================================

baseline_comparison = {

    "category": {

        "accuracy_delta":
            category_metrics[
                "accuracy"
            ]
            -
            BASELINE_RESULTS[
                "category"
            ][
                "accuracy"
            ],

        "macro_f1_delta":
            category_metrics[
                "macro_f1"
            ]
            -
            BASELINE_RESULTS[
                "category"
            ][
                "macro_f1"
            ],
    },

    "subcategory": {

        "accuracy_delta":
            subcategory_metrics[
                "accuracy"
            ]
            -
            BASELINE_RESULTS[
                "subcategory"
            ][
                "accuracy"
            ],

        "macro_f1_delta":
            subcategory_metrics[
                "macro_f1"
            ]
            -
            BASELINE_RESULTS[
                "subcategory"
            ][
                "macro_f1"
            ],
    },

    "priority": {

        "accuracy_delta":
            priority_metrics[
                "accuracy"
            ]
            -
            BASELINE_RESULTS[
                "priority"
            ][
                "accuracy"
            ],

        "macro_f1_delta":
            priority_metrics[
                "macro_f1"
            ]
            -
            BASELINE_RESULTS[
                "priority"
            ][
                "macro_f1"
            ],
    },
}

# ============================================================
# 45. FINAL METRICS JSON
# ============================================================

metrics_output = {

    "experiment":
        "TicketMindAI QLoRA 540 benchmark",

    "n_samples":
        int(
            len(results_df)
        ),

    "valid_json_rate":
        valid_json_rate,

    "malformed_predictions":
        malformed_count,

    "invalid_priority_predictions":
        invalid_priority_count,

    "exact_structured_match_rate":
        exact_structured_match_rate,

    "category":
        category_metrics,

    "subcategory":
        subcategory_metrics,

    "priority":
        priority_metrics,

    "baseline":
        BASELINE_RESULTS,

    "baseline_comparison":
        baseline_comparison,

    "distributions":
        distributions,
}

with open(
    METRICS_PATH,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        metrics_output,
        f,
        indent=2,
        ensure_ascii=False,
    )

# ============================================================
# 46. DISPLAY HELPERS
# ============================================================

def pct(value):

    return (
        f"{value * 100:.2f}%"
    )

def delta_pp(value):

    return (
        f"{value * 100:+.2f} pp"
    )

# ============================================================
# 47. FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("FINAL TICKETMINDAI QLoRA BENCHMARK RESULTS")
print("=" * 70)

print()
print(
    "Samples:",
    len(results_df),
)

print(
    "Valid JSON:",
    pct(
        valid_json_rate
    ),
)

print(
    "Malformed:",
    malformed_count,
)

print(
    "Invalid priority:",
    invalid_priority_count,
)

print(
    "Exact structured match:",
    pct(
        exact_structured_match_rate
    ),
)

# ============================================================
# CATEGORY
# ============================================================

print()
print("## CATEGORY")

print(
    "Accuracy        :",
    pct(
        category_metrics[
            "accuracy"
        ]
    ),
)

print(
    "Macro Precision :",
    f"{category_metrics['macro_precision']:.4f}",
)

print(
    "Macro Recall    :",
    f"{category_metrics['macro_recall']:.4f}",
)

print(
    "Macro F1        :",
    f"{category_metrics['macro_f1']:.4f}",
)

print(
    "Baseline Acc    :",
    pct(
        BASELINE_RESULTS[
            "category"
        ][
            "accuracy"
        ]
    ),
)

print(
    "Accuracy Delta  :",
    delta_pp(
        baseline_comparison[
            "category"
        ][
            "accuracy_delta"
        ]
    ),
)

print(
    "F1 Delta        :",
    delta_pp(
        baseline_comparison[
            "category"
        ][
            "macro_f1_delta"
        ]
    ),
)

# ============================================================
# SUBCATEGORY
# ============================================================

print()
print("## SUBCATEGORY")

print(
    "Accuracy        :",
    pct(
        subcategory_metrics[
            "accuracy"
        ]
    ),
)

print(
    "Macro Precision :",
    f"{subcategory_metrics['macro_precision']:.4f}",
)

print(
    "Macro Recall    :",
    f"{subcategory_metrics['macro_recall']:.4f}",
)

print(
    "Macro F1        :",
    f"{subcategory_metrics['macro_f1']:.4f}",
)

print(
    "Baseline Acc    :",
    pct(
        BASELINE_RESULTS[
            "subcategory"
        ][
            "accuracy"
        ]
    ),
)

print(
    "Accuracy Delta  :",
    delta_pp(
        baseline_comparison[
            "subcategory"
        ][
            "accuracy_delta"
        ]
    ),
)

print(
    "F1 Delta        :",
    delta_pp(
        baseline_comparison[
            "subcategory"
        ][
            "macro_f1_delta"
        ]
    ),
)

# ============================================================
# PRIORITY
# ============================================================

print()
print("## PRIORITY")

print(
    "Accuracy        :",
    pct(
        priority_metrics[
            "accuracy"
        ]
    ),
)

print(
    "Macro Precision :",
    f"{priority_metrics['macro_precision']:.4f}",
)

print(
    "Macro Recall    :",
    f"{priority_metrics['macro_recall']:.4f}",
)

print(
    "Macro F1        :",
    f"{priority_metrics['macro_f1']:.4f}",
)

print(
    "Baseline Acc    :",
    pct(
        BASELINE_RESULTS[
            "priority"
        ][
            "accuracy"
        ]
    ),
)

print(
    "Accuracy Delta  :",
    delta_pp(
        baseline_comparison[
            "priority"
        ][
            "accuracy_delta"
        ]
    ),
)

print(
    "F1 Delta        :",
    delta_pp(
        baseline_comparison[
            "priority"
        ][
            "macro_f1_delta"
        ]
    ),
)

# ============================================================
# 48. FINAL INTEGRITY CHECK
# ============================================================

print()
print("=" * 70)
print("FINAL INTEGRITY CHECK")
print("=" * 70)

assert len(records) == (
    TOTAL_BENCHMARK_ROWS
)

assert len(
    set(
        r["benchmark_index"]
        for r in records
    )
) == TOTAL_BENCHMARK_ROWS

assert not missing_indices

print(
    "✓ Predictions:",
    f"{len(records)}/{TOTAL_BENCHMARK_ROWS}",
)

print(
    "✓ Unique indices:",
    len(all_indices),
)

print(
    "✓ Missing indices:",
    len(missing_indices),
)

# ============================================================
# 49. OUTPUT LOCATIONS
# ============================================================

print()
print("=" * 70)
print("FILES SAVED")
print("=" * 70)

print(
    "Predictions:",
    RAW_PREDICTIONS_PATH,
)

print(
    "Checkpoint:",
    CHECKPOINT_PATH,
)

print(
    "Metrics:",
    METRICS_PATH,
)

print(
    "Errors:",
    ERRORS_PATH,
)

print(
    "Confusion matrices:",
    CONFUSION_DIR,
)

print()
print("=" * 70)
print("✓ TICKETMINDAI 540-TICKET BENCHMARK COMPLETE")
print("=" * 70)
````

````python
# ============================================================
# TICKETMINDAI — PERMANENT 540 EXPERIMENT ARCHIVE
# ============================================================
#
# PURPOSE:
#   Permanently save the COMPLETE 540-ticket experiment.
#
# THIS CELL DOES NOT:
#   - load the model
#   - run inference
#   - consume GPU
#   - regenerate predictions
#
# THIS CELL DOES:
#   - read the 540 checkpoint records
#   - validate all 540 predictions exist
#   - save one complete JSON
#   - save one complete CSV
#   - save metrics
#   - save confusion matrices
#   - save representative errors
#   - create a permanent ZIP archive
#
# AFTER THIS CELL:
#   The 540 inference does NOT need to be run again.
# ============================================================

import os
import json
import shutil
import zipfile
from datetime import datetime

import pandas as pd
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
)

# ============================================================
# 1. PROJECT LOCATION
# ============================================================

DRIVE_ROOT = "/content/drive/MyDrive/TicketMindAI"

if os.path.isdir(DRIVE_ROOT):

    PROJECT_ROOT = DRIVE_ROOT

else:

    PROJECT_ROOT = "/content"

    print(
        "WARNING: Google Drive not detected."
    )

    print(
        "The archive will be stored in /content."
    )

# ============================================================
# 2. MAIN DIRECTORIES
# ============================================================

OUT_DIR = os.path.join(
    PROJECT_ROOT,
    "outputs",
    "ticketmind-qlora",
)

ARCHIVE_DIR = os.path.join(
    OUT_DIR,
    "FINAL_540_EXPERIMENT",
)

CONFUSION_DIR = os.path.join(
    ARCHIVE_DIR,
    "confusion_matrices",
)

os.makedirs(
    CONFUSION_DIR,
    exist_ok=True,
)

# ============================================================
# 3. SOURCE FILES
# ============================================================

CHECKPOINT_PATH = os.path.join(
    OUT_DIR,
    "FINAL_540_checkpoint.jsonl",
)

ADAPTER_DIR = os.path.join(
    OUT_DIR,
    "final_adapter",
)

BENCHMARK_CANDIDATES = [

    os.path.join(
        PROJECT_ROOT,
        "data",
        "processed",
        "benchmark.csv",
    ),

    "/content/data/processed/benchmark.csv",

]

BENCHMARK_PATH = None

for path in BENCHMARK_CANDIDATES:

    if os.path.isfile(path):

        BENCHMARK_PATH = path

        break

if BENCHMARK_PATH is None:

    raise FileNotFoundError(
        "benchmark.csv could not be found."
    )

# ============================================================
# 4. REQUIRED SOURCE VALIDATION
# ============================================================

if not os.path.isfile(
    CHECKPOINT_PATH
):

    raise FileNotFoundError(
        "540 checkpoint does not exist:\n"
        f"{CHECKPOINT_PATH}"
    )

if not os.path.isdir(
    ADAPTER_DIR
):

    raise FileNotFoundError(
        "QLoRA adapter does not exist:\n"
        f"{ADAPTER_DIR}"
    )

if not os.path.isfile(
    os.path.join(
        ADAPTER_DIR,
        "adapter_model.safetensors",
    )
):

    raise FileNotFoundError(
        "adapter_model.safetensors missing."
    )

if not os.path.isfile(
    os.path.join(
        ADAPTER_DIR,
        "adapter_config.json",
    )
):

    raise FileNotFoundError(
        "adapter_config.json missing."
    )

# ============================================================
# 5. LOAD ALL 540 CHECKPOINT RECORDS
# ============================================================

records_by_index = {}

with open(
    CHECKPOINT_PATH,
    "r",
    encoding="utf-8",
) as f:

    for line in f:

        line = line.strip()

        if not line:
            continue

        try:

            record = json.loads(
                line
            )

        except json.JSONDecodeError:

            # Ignore a possible truncated final line.
            continue

        if not isinstance(
            record,
            dict,
        ):

            continue

        if (
            "benchmark_index"
            not in record
        ):

            continue

        idx = int(
            record[
                "benchmark_index"
            ]
        )

        if 0 <= idx < 540:

            records_by_index[idx] = (
                record
            )

# ============================================================
# 6. HARD 540/540 VALIDATION
# ============================================================

expected_indices = set(
    range(540)
)

actual_indices = set(
    records_by_index.keys()
)

missing_indices = sorted(
    expected_indices
    -
    actual_indices
)

extra_indices = sorted(
    actual_indices
    -
    expected_indices
)

print()
print("=" * 70)
print("540 CHECKPOINT VALIDATION")
print("=" * 70)

print(
    "Expected records :",
    540,
)

print(
    "Found records    :",
    len(records_by_index),
)

print(
    "Missing records  :",
    len(missing_indices),
)

print(
    "Extra records    :",
    len(extra_indices),
)

if missing_indices:

    raise RuntimeError(
        "STOP.\n"
        "The 540 benchmark is incomplete.\n\n"
        f"Missing indices:\n{missing_indices}"
    )

if extra_indices:

    raise RuntimeError(
        "Unexpected benchmark indices:\n"
        f"{extra_indices}"
    )

assert len(
    records_by_index
) == 540

print(
    "✓ ALL 540 PREDICTIONS EXIST"
)

# ============================================================
# 7. SORT IN ORIGINAL BENCHMARK ORDER
# ============================================================

records = [

    records_by_index[i]

    for i in range(540)

]

results = pd.DataFrame(
    records
)

assert len(results) == 540

# ============================================================
# 8. LOAD ORIGINAL BENCHMARK
# ============================================================

benchmark_df = pd.read_csv(
    BENCHMARK_PATH
).reset_index(drop=True)

if len(
    benchmark_df
) != 540:

    raise RuntimeError(
        "Benchmark must contain exactly "
        f"540 rows. Found {len(benchmark_df)}."
    )

# ============================================================
# 9. VERIFY BENCHMARK ORDER
# ============================================================

for i in range(540):

    checkpoint_input = str(
        records[i]["input"]
    ).strip()

    benchmark_input = str(
        benchmark_df.iloc[i]["input"]
    ).strip()

    if checkpoint_input != (
        benchmark_input
    ):

        raise RuntimeError(
            "BENCHMARK ORDER MISMATCH.\n"
            f"Index: {i}\n"
            f"Checkpoint: {checkpoint_input}\n"
            f"Benchmark: {benchmark_input}"
        )

print(
    "✓ Benchmark order verified"
)

# ============================================================
# 10. CREATE COMPLETE 540 DATASET
# ============================================================
#
# This is the most important file.
#
# Every row contains:
#
# INPUT
# TRUE LABELS
# RAW MODEL OUTPUT
# PARSED PREDICTIONS
# VALIDITY FLAGS
#
# ============================================================

complete_540 = []

for record in records:

    complete_540.append({

        "benchmark_index":
            record[
                "benchmark_index"
            ],

        "input":
            record["input"],

        "true_category":
            record[
                "true_category"
            ],

        "pred_category":
            record[
                "pred_category"
            ],

        "true_subcategory":
            record[
                "true_subcategory"
            ],

        "pred_subcategory":
            record[
                "pred_subcategory"
            ],

        "true_priority":
            record[
                "true_priority"
            ],

        "pred_priority_raw":
            record[
                "pred_priority_raw"
            ],

        "pred_priority_norm":
            record[
                "pred_priority_norm"
            ],

        "priority_valid":
            record[
                "priority_valid"
            ],

        "pred_sentiment":
            record[
                "pred_sentiment"
            ],

        "pred_summary":
            record[
                "pred_summary"
            ],

        "valid_json":
            record[
                "valid_json"
            ],

        "malformed":
            record[
                "malformed"
            ],

        "raw_model_output":
            record[
                "raw_model_output"
            ],
    })

complete_df = pd.DataFrame(
    complete_540
)

assert len(
    complete_df
) == 540

# ============================================================
# 11. SAVE COMPLETE JSON
# ============================================================

COMPLETE_JSON = os.path.join(
    ARCHIVE_DIR,
    "FINAL_540_COMPLETE.json",
)

with open(
    COMPLETE_JSON,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        complete_540,
        f,
        indent=2,
        ensure_ascii=False,
    )

print(
    "✓ Complete JSON saved"
)

# ============================================================
# 12. SAVE COMPLETE CSV
# ============================================================

COMPLETE_CSV = os.path.join(
    ARCHIVE_DIR,
    "FINAL_540_COMPLETE.csv",
)

complete_df.to_csv(
    COMPLETE_CSV,
    index=False,
    encoding="utf-8-sig",
)

print(
    "✓ Complete CSV saved"
)

# ============================================================
# 13. METRICS
# ============================================================

category_labels = sorted(
    complete_df[
        "true_category"
    ].unique()
)

subcategory_labels = sorted(
    complete_df[
        "true_subcategory"
    ].unique()
)

priority_labels = [
    "HIGH",
    "MEDIUM",
    "LOW",
]

def calculate_metrics(
    true,
    pred,
    labels,
):

    accuracy = accuracy_score(
        true,
        pred,
    )

    precision, recall, f1, _ = (
        precision_recall_fscore_support(
            true,
            pred,
            labels=labels,
            average="macro",
            zero_division=0,
        )
    )

    return {

        "accuracy":
            float(accuracy),

        "macro_precision":
            float(precision),

        "macro_recall":
            float(recall),

        "macro_f1":
            float(f1),
    }

# ============================================================
# 14. CATEGORY METRICS
# ============================================================

category_metrics = calculate_metrics(

    complete_df[
        "true_category"
    ],

    complete_df[
        "pred_category"
    ],

    category_labels,
)

# ============================================================
# 15. SUBCATEGORY METRICS
# ============================================================

subcategory_metrics = calculate_metrics(

    complete_df[
        "true_subcategory"
    ],

    complete_df[
        "pred_subcategory"
    ],

    subcategory_labels,
)

# ============================================================
# 16. PRIORITY METRICS
# ============================================================

priority_metrics = calculate_metrics(

    complete_df[
        "true_priority"
    ],

    complete_df[
        "pred_priority_norm"
    ],

    priority_labels,
)

# ============================================================
# 17. JSON QUALITY
# ============================================================

valid_json_rate = float(
    complete_df[
        "valid_json"
    ].mean()
)

malformed_count = int(
    complete_df[
        "malformed"
    ].sum()
)

invalid_priority_count = int(
    (
        ~complete_df[
            "priority_valid"
        ]
    ).sum()
)

# ============================================================
# 18. EXACT STRUCTURED MATCH
# ============================================================

exact_match = (

    (
        complete_df[
            "true_category"
        ]
        ==
        complete_df[
            "pred_category"
        ]
    )

    &

    (
        complete_df[
            "true_subcategory"
        ]
        ==
        complete_df[
            "pred_subcategory"
        ]
    )

    &

    (
        complete_df[
            "true_priority"
        ]
        ==
        complete_df[
            "pred_priority_norm"
        ]
    )
)

exact_match_rate = float(
    exact_match.mean()
)

# ============================================================
# 19. PROMPT BASELINE
# ============================================================

BASELINE = {

    "category": {

        "accuracy":
            0.9741,

        "macro_precision":
            0.9762,

        "macro_recall":
            0.9708,

        "macro_f1":
            0.9716,
    },

    "subcategory": {

        "accuracy":
            0.9481,

        "macro_precision":
            0.9537,

        "macro_recall":
            0.9481,

        "macro_f1":
            0.9480,
    },

    "priority": {

        "accuracy":
            0.3870,

        "macro_precision":
            0.2176,

        "macro_recall":
            0.3653,

        "macro_f1":
            0.2711,
    },
}

# ============================================================
# 20. BASELINE COMPARISON
# ============================================================

comparison = {

    "category": {

        "accuracy_delta":
            category_metrics[
                "accuracy"
            ]
            -
            BASELINE[
                "category"
            ][
                "accuracy"
            ],

        "macro_f1_delta":
            category_metrics[
                "macro_f1"
            ]
            -
            BASELINE[
                "category"
            ][
                "macro_f1"
            ],
    },

    "subcategory": {

        "accuracy_delta":
            subcategory_metrics[
                "accuracy"
            ]
            -
            BASELINE[
                "subcategory"
            ][
                "accuracy"
            ],

        "macro_f1_delta":
            subcategory_metrics[
                "macro_f1"
            ]
            -
            BASELINE[
                "subcategory"
            ][
                "macro_f1"
            ],
    },

    "priority": {

        "accuracy_delta":
            priority_metrics[
                "accuracy"
            ]
            -
            BASELINE[
                "priority"
            ][
                "accuracy"
            ],

        "macro_f1_delta":
            priority_metrics[
                "macro_f1"
            ]
            -
            BASELINE[
                "priority"
            ][
                "macro_f1"
            ],
    },
}

# ============================================================
# 21. CONFUSION MATRICES
# ============================================================

category_cm = confusion_matrix(

    complete_df[
        "true_category"
    ],

    complete_df[
        "pred_category"
    ],

    labels=category_labels,
)

pd.DataFrame(

    category_cm,

    index=category_labels,

    columns=category_labels,

).to_csv(

    os.path.join(
        CONFUSION_DIR,
        "category_confusion.csv",
    )
)

subcategory_cm = confusion_matrix(

    complete_df[
        "true_subcategory"
    ],

    complete_df[
        "pred_subcategory"
    ],

    labels=subcategory_labels,
)

pd.DataFrame(

    subcategory_cm,

    index=subcategory_labels,

    columns=subcategory_labels,

).to_csv(

    os.path.join(
        CONFUSION_DIR,
        "subcategory_confusion.csv",
    )
)

priority_cm_labels = [

    "HIGH",
    "MEDIUM",
    "LOW",
    "INVALID",
]

priority_cm = confusion_matrix(

    complete_df[
        "true_priority"
    ],

    complete_df[
        "pred_priority_norm"
    ],

    labels=priority_cm_labels,
)

pd.DataFrame(

    priority_cm,

    index=priority_cm_labels,

    columns=priority_cm_labels,

).to_csv(

    os.path.join(
        CONFUSION_DIR,
        "priority_confusion.csv",
    )
)

# ============================================================
# 22. REPRESENTATIVE ERRORS
# ============================================================

errors = complete_df[
    ~exact_match
].copy()

if len(errors) > 15:

    representative_errors = (
        errors.sample(
            n=15,
            random_state=42,
        )
    )

else:

    representative_errors = (
        errors
    )

ERRORS_CSV = os.path.join(
    ARCHIVE_DIR,
    "FINAL_540_REPRESENTATIVE_ERRORS.csv",
)

representative_errors.to_csv(
    ERRORS_CSV,
    index=False,
)

# ============================================================
# 23. DISTRIBUTIONS
# ============================================================

distributions = {

    "true_category":
        complete_df[
            "true_category"
        ]
        .value_counts()
        .to_dict(),

    "pred_category":
        complete_df[
            "pred_category"
        ]
        .value_counts()
        .to_dict(),

    "true_subcategory":
        complete_df[
            "true_subcategory"
        ]
        .value_counts()
        .to_dict(),

    "pred_subcategory":
        complete_df[
            "pred_subcategory"
        ]
        .value_counts()
        .to_dict(),

    "true_priority":
        complete_df[
            "true_priority"
        ]
        .value_counts()
        .to_dict(),

    "pred_priority_raw":
        complete_df[
            "pred_priority_raw"
        ]
        .value_counts()
        .to_dict(),

    "pred_priority_norm":
        complete_df[
            "pred_priority_norm"
        ]
        .value_counts()
        .to_dict(),
}

# ============================================================
# 24. METRICS JSON
# ============================================================

final_metrics = {

    "experiment":
        "TicketMindAI FINAL QLoRA 540 Benchmark",

    "n_samples":
        540,

    "benchmark_file":
        BENCHMARK_PATH,

    "adapter_directory":
        ADAPTER_DIR,

    "valid_json_rate":
        valid_json_rate,

    "malformed_predictions":
        malformed_count,

    "invalid_priority_predictions":
        invalid_priority_count,

    "exact_structured_match_rate":
        exact_match_rate,

    "category":
        category_metrics,

    "subcategory":
        subcategory_metrics,

    "priority":
        priority_metrics,

    "baseline":
        BASELINE,

    "baseline_comparison":
        comparison,

    "distributions":
        distributions,
}

METRICS_JSON = os.path.join(
    ARCHIVE_DIR,
    "FINAL_540_METRICS.json",
)

with open(
    METRICS_JSON,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        final_metrics,
        f,
        indent=2,
        ensure_ascii=False,
    )

# ============================================================
# 25. COPY ORIGINAL BENCHMARK
# ============================================================

ARCHIVED_BENCHMARK = os.path.join(
    ARCHIVE_DIR,
    "benchmark_540_original.csv",
)

shutil.copy2(
    BENCHMARK_PATH,
    ARCHIVED_BENCHMARK,
)

print(
    "✓ Original benchmark archived"
)

# ============================================================
# 26. COPY CHECKPOINT
# ============================================================

ARCHIVED_CHECKPOINT = os.path.join(
    ARCHIVE_DIR,
    "FINAL_540_checkpoint.jsonl",
)

shutil.copy2(
    CHECKPOINT_PATH,
    ARCHIVED_CHECKPOINT,
)

print(
    "✓ Checkpoint archived"
)

# ============================================================
# 27. COPY QLORA ADAPTER
# ============================================================

ARCHIVED_ADAPTER = os.path.join(
    ARCHIVE_DIR,
    "final_adapter",
)

if os.path.exists(
    ARCHIVED_ADAPTER
):

    shutil.rmtree(
        ARCHIVED_ADAPTER
    )

shutil.copytree(
    ADAPTER_DIR,
    ARCHIVED_ADAPTER,
)

print(
    "✓ Trained QLoRA adapter archived"
)

# ============================================================
# 28. CREATE MANIFEST
# ============================================================

manifest = {

    "project":
        "TicketMindAI",

    "experiment":
        "FINAL QLoRA 540 benchmark",

    "created_at":
        datetime.now().isoformat(),

    "samples":
        540,

    "benchmark":
        "benchmark_540_original.csv",

    "trained_adapter":
        "final_adapter/",

    "complete_predictions":
        "FINAL_540_COMPLETE.json",

    "complete_predictions_csv":
        "FINAL_540_COMPLETE.csv",

    "metrics":
        "FINAL_540_METRICS.json",

    "representative_errors":
        "FINAL_540_REPRESENTATIVE_ERRORS.csv",

    "checkpoint":
        "FINAL_540_checkpoint.jsonl",

    "confusion_matrices":
        "confusion_matrices/",

    "inference_required_after_archive":
        False,

    "benchmark_rerun_required":
        False,

    "training_rerun_required":
        False,

    "notes": [
        "540 benchmark predictions are permanently archived.",
        "Raw model outputs are preserved.",
        "Ground-truth labels are preserved.",
        "Parsed predictions are preserved.",
        "QLoRA adapter is preserved.",
        "Evaluation metrics are preserved.",
        "Confusion matrices are preserved.",
        "The archive is sufficient to reproduce the reported evaluation analysis without rerunning model inference."
    ],
}

MANIFEST_PATH = os.path.join(
    ARCHIVE_DIR,
    "MANIFEST.json",
)

with open(
    MANIFEST_PATH,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        manifest,
        f,
        indent=2,
        ensure_ascii=False,
    )

print(
    "✓ Manifest created"
)

# ============================================================
# 29. CREATE PERMANENT ZIP
# ============================================================
#
# This gives you ONE portable file.
#
# It contains:
#   - trained adapter
#   - all 540 predictions
#   - original benchmark
#   - metrics
#   - checkpoint
#   - confusion matrices
#   - errors
#   - manifest
#
# ============================================================

ZIP_PATH = os.path.join(
    OUT_DIR,
    "TicketMindAI_FINAL_540_EXPERIMENT.zip",
)

if os.path.isfile(
    ZIP_PATH
):

    os.remove(
        ZIP_PATH
    )

with zipfile.ZipFile(
    ZIP_PATH,
    "w",
    compression=zipfile.ZIP_DEFLATED,
) as zipf:

    for root, dirs, files in os.walk(
        ARCHIVE_DIR
    ):

        for filename in files:

            full_path = os.path.join(
                root,
                filename,
            )

            relative_path = os.path.relpath(
                full_path,
                ARCHIVE_DIR,
            )

            zipf.write(
                full_path,
                arcname=relative_path,
            )

# ============================================================
# 30. VERIFY ZIP
# ============================================================

with zipfile.ZipFile(
    ZIP_PATH,
    "r",
) as zipf:

    bad_file = zipf.testzip()

    if bad_file is not None:

        raise RuntimeError(
            "ZIP verification failed: "
            f"{bad_file}"
        )

print(
    "✓ ZIP archive verified"
)

# ============================================================
# 31. FINAL REPORT
# ============================================================

def pct(x):

    return (
        f"{x * 100:.2f}%"
    )

def delta(x):

    return (
        f"{x * 100:+.2f} pp"
    )

print()
print("=" * 70)
print("TICKETMINDAI — PERMANENT EXPERIMENT SAVED")
print("=" * 70)

print()
print(
    "540 predictions        : SAVED"
)

print(
    "Raw model outputs      : SAVED"
)

print(
    "Ground truth           : SAVED"
)

print(
    "QLoRA adapter          : SAVED"
)

print(
    "Metrics                : SAVED"
)

print(
    "Confusion matrices     : SAVED"
)

print(
    "Representative errors  : SAVED"
)

print(
    "Checkpoint             : SAVED"
)

print()
print("=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print()
print(
    "Category Accuracy      :",
    pct(
        category_metrics[
            "accuracy"
        ]
    ),
)

print(
    "Category Macro F1      :",
    f"{category_metrics['macro_f1']:.4f}",
)

print(
    "Subcategory Accuracy   :",
    pct(
        subcategory_metrics[
            "accuracy"
        ]
    ),
)

print(
    "Subcategory Macro F1   :",
    f"{subcategory_metrics['macro_f1']:.4f}",
)

print(
    "Priority Accuracy      :",
    pct(
        priority_metrics[
            "accuracy"
        ]
    ),
)

print(
    "Priority Macro F1      :",
    f"{priority_metrics['macro_f1']:.4f}",
)

print(
    "Valid JSON             :",
    pct(
        valid_json_rate
    ),
)

print(
    "Exact Structured Match :",
    pct(
        exact_match_rate
    ),
)

print()
print("=" * 70)
print("IMPORTANT FILES")
print("=" * 70)

print()
print(
    "COMPLETE EXPERIMENT:"
)

print(
    ARCHIVE_DIR
)

print()
print(
    "ONE-FILE PORTABLE ARCHIVE:"
)

print(
    ZIP_PATH
)

print()
print(
    "COMPLETE 540 JSON:"
)

print(
    COMPLETE_JSON
)

print()
print(
    "COMPLETE 540 CSV:"
)

print(
    COMPLETE_CSV
)

print()
print("=" * 70)
print("✓ 540/540 PERMANENTLY ARCHIVED")
print("✓ TRAINED ADAPTER PERMANENTLY ARCHIVED")
print("✓ NO FUTURE 540 INFERENCE REQUIRED")
print("✓ NO FUTURE TRAINING REQUIRED")
print("=" * 70)
````

````python
# ============================================================
# TICKETMINDAI — SAVED 540 TESTCASE AUDIT
# NO MODEL / NO GPU / NO INFERENCE
# ============================================================

import os
import json
import pandas as pd
from collections import Counter

ARCHIVE = (
    "/content/drive/MyDrive/TicketMindAI/"
    "outputs/ticketmind-qlora/FINAL_540_EXPERIMENT"
)

JSON_PATH = os.path.join(
    ARCHIVE,
    "FINAL_540_COMPLETE.json"
)

CSV_PATH = os.path.join(
    ARCHIVE,
    "FINAL_540_COMPLETE.csv"
)

RESULTS_PATH = os.path.join(
    ARCHIVE,
    "SAVED_540_TEST_RESULTS.json"
)

REPORT_PATH = os.path.join(
    ARCHIVE,
    "SAVED_540_TEST_REPORT.txt"
)

# ============================================================
# LOAD
# ============================================================

if not os.path.exists(JSON_PATH):
    raise FileNotFoundError(JSON_PATH)

with open(
    JSON_PATH,
    "r",
    encoding="utf-8"
) as f:
    data = json.load(f)

assert len(data) == 540

print("=" * 70)
print("TICKETMINDAI — SAVED 540 TESTCASE AUDIT")
print("=" * 70)

print("Loaded:", len(data), "testcases")

# ============================================================
# TEST 1 — INDEX INTEGRITY
# ============================================================

indices = [
    x["benchmark_index"]
    for x in data
]

expected_indices = list(range(540))

index_test = (
    indices == expected_indices
)

# ============================================================
# TEST 2 — REQUIRED FIELDS
# ============================================================

required_fields = [
    "benchmark_index",
    "input",
    "true_category",
    "pred_category",
    "true_subcategory",
    "pred_subcategory",
    "true_priority",
    "pred_priority_raw",
    "pred_priority_norm",
    "priority_valid",
    "pred_sentiment",
    "pred_summary",
    "valid_json",
    "malformed",
    "raw_model_output",
]

missing_fields = []

for i, item in enumerate(data):

    for field in required_fields:

        if field not in item:

            missing_fields.append(
                (i, field)
            )

# ============================================================
# TEST 3 — RAW OUTPUT EXISTENCE
# ============================================================

empty_raw = []

for i, item in enumerate(data):

    raw = item.get(
        "raw_model_output"
    )

    if (
        raw is None
        or not str(raw).strip()
    ):

        empty_raw.append(i)

# ============================================================
# TEST 4 — JSON VALIDITY
# ============================================================

invalid_json_indices = [
    i
    for i, x in enumerate(data)
    if not x.get("valid_json", False)
]

# ============================================================
# TEST 5 — CATEGORY
# ============================================================

category_correct = [
    x["pred_category"]
    ==
    x["true_category"]
    for x in data
]

category_accuracy = (
    sum(category_correct)
    / 540
)

# ============================================================
# TEST 6 — SUBCATEGORY
# ============================================================

subcategory_correct = [
    x["pred_subcategory"]
    ==
    x["true_subcategory"]
    for x in data
]

subcategory_accuracy = (
    sum(subcategory_correct)
    / 540
)

# ============================================================
# TEST 7 — PRIORITY
# ============================================================

priority_correct = [
    x["pred_priority_norm"]
    ==
    x["true_priority"]
    for x in data
]

priority_accuracy = (
    sum(priority_correct)
    / 540
)

# ============================================================
# TEST 8 — COMPLETE STRUCTURED MATCH
# ============================================================

exact_correct = []

for x in data:

    match = (

        x["pred_category"]
        ==
        x["true_category"]

        and

        x["pred_subcategory"]
        ==
        x["true_subcategory"]

        and

        x["pred_priority_norm"]
        ==
        x["true_priority"]

    )

    exact_correct.append(match)

exact_accuracy = (
    sum(exact_correct)
    / 540
)

# ============================================================
# TEST 9 — PRIORITY VALIDITY
# ============================================================

valid_priorities = {
    "HIGH",
    "MEDIUM",
    "LOW"
}

invalid_priority_indices = []

for i, x in enumerate(data):

    if (
        x["pred_priority_norm"]
        not in valid_priorities
    ):

        invalid_priority_indices.append(i)

# ============================================================
# TEST 10 — PREDICTION DISTRIBUTIONS
# ============================================================

true_categories = Counter(
    x["true_category"]
    for x in data
)

pred_categories = Counter(
    x["pred_category"]
    for x in data
)

true_subcategories = Counter(
    x["true_subcategory"]
    for x in data
)

pred_subcategories = Counter(
    x["pred_subcategory"]
    for x in data
)

true_priorities = Counter(
    x["true_priority"]
    for x in data
)

pred_priorities = Counter(
    x["pred_priority_norm"]
    for x in data
)

# ============================================================
# TEST 11 — FIND ALL ERRORS
# ============================================================

category_errors = [
    i
    for i, x in enumerate(data)
    if x["pred_category"]
    !=
    x["true_category"]
]

subcategory_errors = [
    i
    for i, x in enumerate(data)
    if x["pred_subcategory"]
    !=
    x["true_subcategory"]
]

priority_errors = [
    i
    for i, x in enumerate(data)
    if x["pred_priority_norm"]
    !=
    x["true_priority"]
]

exact_errors = [
    i
    for i, x in enumerate(data)
    if not (
        x["pred_category"]
        ==
        x["true_category"]

        and

        x["pred_subcategory"]
        ==
        x["true_subcategory"]

        and

        x["pred_priority_norm"]
        ==
        x["true_priority"]
    )
]

# ============================================================
# TEST 12 — RAW OUTPUT SANITY
# ============================================================

raw_contains_prediction = 0

for x in data:

    raw = str(
        x["raw_model_output"]
    )

    pred_category = str(
        x["pred_category"]
    )

    pred_subcategory = str(
        x["pred_subcategory"]
    )

    if (
        pred_category in raw
        and
        pred_subcategory in raw
    ):

        raw_contains_prediction += 1

raw_prediction_rate = (
    raw_contains_prediction
    / 540
)

# ============================================================
# PRINT RESULTS
# ============================================================

print()
print("=" * 70)
print("TEST RESULTS")
print("=" * 70)

print()
print(
    "1. Record count             :",
    "PASS" if len(data) == 540 else "FAIL"
)

print(
    "2. Index integrity          :",
    "PASS" if index_test else "FAIL"
)

print(
    "3. Required fields          :",
    "PASS" if not missing_fields else "FAIL"
)

print(
    "4. Raw outputs present      :",
    "PASS" if not empty_raw else "FAIL"
)

print(
    "5. Valid JSON               :",
    "PASS" if not invalid_json_indices else "FAIL"
)

print(
    "6. Category                 :",
    f"{category_accuracy * 100:.2f}%"
)

print(
    "7. Subcategory              :",
    f"{subcategory_accuracy * 100:.2f}%"
)

print(
    "8. Priority                 :",
    f"{priority_accuracy * 100:.2f}%"
)

print(
    "9. Exact structured match   :",
    f"{exact_accuracy * 100:.2f}%"
)

print(
    "10. Priority validity       :",
    "PASS" if not invalid_priority_indices else "FAIL"
)

print(
    "11. Raw-output consistency  :",
    f"{raw_prediction_rate * 100:.2f}%"
)

# ============================================================
# ERROR COUNTS
# ============================================================

print()
print("=" * 70)
print("ERROR COUNTS")
print("=" * 70)

print(
    "Category errors      :",
    len(category_errors)
)

print(
    "Subcategory errors   :",
    len(subcategory_errors)
)

print(
    "Priority errors      :",
    len(priority_errors)
)

print(
    "Exact-match errors   :",
    len(exact_errors)
)

print(
    "Invalid JSON         :",
    len(invalid_json_indices)
)

print(
    "Invalid priorities   :",
    len(invalid_priority_indices)
)

# ============================================================
# SHOW ERRORS IF ANY
# ============================================================

if exact_errors:

    print()
    print("=" * 70)
    print("FIRST 20 ERRORS")
    print("=" * 70)

    for idx in exact_errors[:20]:

        x = data[idx]

        print()
        print("---", idx, "---")

        print(
            "INPUT:",
            x["input"]
        )

        print(
            "TRUE:",
            x["true_category"],
            "|",
            x["true_subcategory"],
            "|",
            x["true_priority"]
        )

        print(
            "PRED:",
            x["pred_category"],
            "|",
            x["pred_subcategory"],
            "|",
            x["pred_priority_norm"]
        )

        print(
            "RAW:",
            x["raw_model_output"]
        )

else:

    print()
    print(
        "✓ NO STRUCTURED PREDICTION ERRORS"
    )

# ============================================================
# SAVE RESULTS
# ============================================================

results = {

    "test_name":
        "TicketMindAI Saved 540 Testcase Audit",

    "samples":
        540,

    "record_count":
        len(data),

    "index_integrity":
        index_test,

    "missing_required_fields":
        len(missing_fields),

    "empty_raw_outputs":
        len(empty_raw),

    "invalid_json":
        len(invalid_json_indices),

    "invalid_priorities":
        len(invalid_priority_indices),

    "category_accuracy":
        category_accuracy,

    "subcategory_accuracy":
        subcategory_accuracy,

    "priority_accuracy":
        priority_accuracy,

    "exact_structured_match":
        exact_accuracy,

    "raw_output_prediction_consistency":
        raw_prediction_rate,

    "category_errors":
        len(category_errors),

    "subcategory_errors":
        len(subcategory_errors),

    "priority_errors":
        len(priority_errors),

    "exact_match_errors":
        len(exact_errors),

    "true_category_distribution":
        dict(true_categories),

    "pred_category_distribution":
        dict(pred_categories),

    "true_subcategory_distribution":
        dict(true_subcategories),

    "pred_subcategory_distribution":
        dict(pred_subcategories),

    "true_priority_distribution":
        dict(true_priorities),

    "pred_priority_distribution":
        dict(pred_priorities),

    "status":
        (
            "PASS"
            if (
                len(data) == 540
                and index_test
                and not missing_fields
                and not empty_raw
                and not invalid_json_indices
                and not invalid_priority_indices
            )
            else "CHECK REQUIRED"
        )
}

with open(
    RESULTS_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        results,
        f,
        indent=2,
        ensure_ascii=False
    )

# ============================================================
# HUMAN-READABLE REPORT
# ============================================================

report_lines = [

    "TICKETMINDAI — SAVED 540 TESTCASE AUDIT",
    "=" * 70,
    "",
    f"Samples: {len(data)}",
    f"Index integrity: {index_test}",
    f"Missing fields: {len(missing_fields)}",
    f"Empty raw outputs: {len(empty_raw)}",
    f"Invalid JSON: {len(invalid_json_indices)}",
    f"Invalid priorities: {len(invalid_priority_indices)}",
    "",
    f"Category Accuracy: {category_accuracy:.6f}",
    f"Subcategory Accuracy: {subcategory_accuracy:.6f}",
    f"Priority Accuracy: {priority_accuracy:.6f}",
    f"Exact Structured Match: {exact_accuracy:.6f}",
    f"Raw Output Consistency: {raw_prediction_rate:.6f}",
    "",
    f"Category Errors: {len(category_errors)}",
    f"Subcategory Errors: {len(subcategory_errors)}",
    f"Priority Errors: {len(priority_errors)}",
    f"Exact Match Errors: {len(exact_errors)}",
    "",
    "STATUS: "
    + (
        "PASS"
        if results["status"] == "PASS"
        else "CHECK REQUIRED"
    ),
]

with open(
    REPORT_PATH,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "\n".join(
            report_lines
        )
    )

# ============================================================
# FINAL
# ============================================================

print()
print("=" * 70)
print("SAVED TESTCASE AUDIT COMPLETE")
print("=" * 70)

print(
    "Results saved:",
    RESULTS_PATH
)

print(
    "Report saved :",
    REPORT_PATH
)

print()
print(
    "✓ NO MODEL WAS RUN"
)

print(
    "✓ NO GPU WAS USED"
)

print(
    "✓ ALL 540 SAVED TESTCASES WERE TESTED"
)

print("=" * 70)
````
